```python
import pandas as pd

df_titanic = pd.read_csv("titanic.xls")   # naam .xls hai but andar CSV hai
df_students = pd.read_csv("StudentsPerformance.csv")
df_tips = pd.read_csv("tips.csv")

print(df_titanic.head())
print(df_students.head())
print(df_tips.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='Survived', data=df_titanic)
plt.title("Survival Count")
plt.show()
```

```python

survival_rate = df_titanic.groupby('Sex')['Survived'].mean()
print(survival_rate)
```

```python
print("Highest survival rate gender:", survival_rate.idxmax())

lowest_age_class = df_titanic.groupby('Pclass')['Age'].mean().idxmin()
print("Lowest avg age class:", lowest_age_class)
```

```python
print(df_students.head())
```

```python
print(df_students['gender'].value_counts())
```

```python
print("Math avg:", df_students['math score'].mean())
print("Reading avg:", df_students['reading score'].mean())
print("Writing avg:", df_students['writing score'].mean())
```

```python
df_students['Total'] = df_students['math score'] + df_students['reading score'] + df_students['writing score']
print(df_students.loc[df_students['Total'].idxmax()])
```

```python
df_students['Percentage'] = df_students['Total'] / 3
print(df_students.head())
```

```python
high_scorers = df_students[(df_students['math score'] > 70) & (df_students['reading score'] > 70) & (df_students['writing score'] > 70)]
print(high_scorers)
print("Total such students:", len(high_scorers))
```

```python
print(df_students.groupby('gender')[['math score','reading score','writing score']].mean())
```

```python
df_sorted = df_students.sort_values(by='Total', ascending=False)
print(df_sorted.head())
```

```python
top10 = df_students.sort_values(by='Percentage', ascending=False).head(10)
print(top10)
```

```python
pivot = df_students.pivot_table(index='gender', values='math score', aggfunc='mean')
print(pivot)
```
